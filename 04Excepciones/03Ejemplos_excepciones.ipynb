# Ejemplos excepciones

## Con funciones

```python
def pide_entero() -> int | None:
    valor = input('Escribe un número natural: ')
    try:
        return int(valor)
    except ValueError:
        print(valor, 'no es un número entero válido.')
        return None
```

## Con bucle

```python
valor_correcto = False

while not valor_correcto:
    try:
        valor = input('Escribe un número natural: ')
        print('El recíproco de', valor, 'es', 1/int(valor))        
        valor_correcto = True
    except ValueError:
        print(valor, 'no es un número entero válido.')    
    except ZeroDivisionError:
        print('La división entre cero no está permitida en nuestro Universo.')
```
