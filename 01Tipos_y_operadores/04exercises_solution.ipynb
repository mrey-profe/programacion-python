```python
# Declare variables
print("Imos calcular a área dun triángulo")
base = int(input("Escribe a base: "))
height = 5

# Calculate the area
area = base * height / 2

# Print the result
print("The area of a triangle with base", base, "and height", height, "is", area)
```

```python
# Declare variables
base = 2
height = 5

# Calculate the area
area = base * height
perimeter = 2 * (base + height)

# Print the result
print("The area of a rectangle with base", base, "and height", height, "is", area)
print("The perimeter of a rectangle with base", base, "and height", height, "is", perimeter)
```
