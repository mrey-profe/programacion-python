## Exercise 1

Create a function that receives a number and returns True if the number is even and False if it is odd.

Then, the program should ask for two numbers and display the even numbers between them.

```python
def is_even(number: int) -> bool:
    if number % 2 == 0:
        return True
    else:
        return False

# Main program
if is_even(2):
    print("Even number")
else:
    print("Odd number")
```

## Exercise 3

Create a function that receives a number and returns the factorial of that number.

```python
def factorial(n: int) -> int:
    '''
    >>> factorial(5)
    120
    >>> factorial(4)
    24
    >>> factorial(0)
    1
    '''
    resultado = 1
    
    for i in range(n, 1, -1): #range(1,n+1)
        resultado *= i
        
    return resultado
```

```python
def factorial(n: int) -> int|None:
    '''
    >>> factorial(5)
    120
    >>> factorial(4)
    24
    >>> factorial(0)
    1
    >>> factorial(-3)
    
    '''
    if n < 0:
        return None
    
    resultado = 1
    
    for i in range(n, 1, -1): #range(1,n+1)
        resultado *= i
        
    return resultado
```
