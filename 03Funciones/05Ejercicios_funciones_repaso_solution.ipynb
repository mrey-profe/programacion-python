## Exercise 2

Create a function that converts degrees Celsius to Fahrenheit. It should receive a number in Celsius and return its equivalent in Fahrenheit.

```python
def celsius_to_fahrenheit(celsius: float) -> float:
    '''
    Convert Celsius to Fahrenheit.
    
    Parameters:
    -----------
    celsius (float): Temperature in Celsius.
    Returns:
    --------
    float: Temperature in Fahrenheit.
    >>> celsius_to_fahrenheit(0)
    32.0
    >>> celsius_to_fahrenheit(100)
    212.0'''
    return (celsius * 9/5) + 32
```

## Exercise 3

Create a function that receives a number and returns the factorial of that number.

```python
def factorial(n: int) -> int:
    '''
    >>> factorial(5)
    120
    >>> factorial(4)
    24
    >>> factorial(0)
    1
    '''
    resultado = 1
    
    for i in range(n, 1, -1): #range(1,n+1)
        resultado *= i
        
    return resultado
```

```python
def factorial(n: int) -> int|None:
    '''
    >>> factorial(5)
    120
    >>> factorial(4)
    24
    >>> factorial(0)
    1
    >>> factorial(-3)
    
    '''
    if n < 0:
        return None
    
    resultado = 1
    
    for i in range(n, 1, -1): #range(1,n+1)
        resultado *= i
        
    return resultado
```

## Exercise 4

Create a function that returns a random number within a given range. The function should receive `min` and `max` as arguments, with default values 1 and 100, respectively.

```python
import random

def random(min: int, max: int) -> int:
    '''
    Return a random integer between min and max (inclusive).
    
    Parameters:
    -----------
    min (int): Minimum value.
    max (int): Maximum value.
    
    Returns:
    --------
    int: Random integer between min and max.
    '''
    return random.randint(min, max)
```

## Exercise 5

Create a function that counts how many vowels are in a text string. The function should receive a string and return the number of vowels.

```python
def count_vowels(s: str) -> int:
    '''
    Count the number of vowels in a string.
    
    Parameters:
    -----------
    s (str): Input string.
    
    Returns:
    --------
    int: Number of vowels in the string.
    
    >>> count_vowels("Hello World")
    3
    >>> count_vowels("Python")
    1
    >>> count_vowels("AEIOU")
    5
    '''
    vowels = "AEIOUÁÉÍÓÚÜ"
    counter = 0
    for char in s.upper():
        if char in vowels:
            counter += 1
    return counter
```

## Exercise 6

Create a function that classifies a person as "child", "teenager", "adult" or "elderly" based on their age. The function should accept an `age` argument and return a string with the classification.

```python
def age_classify(age: int) -> str:
    '''
    Classify a person based on their age.
    
    Parameters:
    -----------
    age (int): Age of the person.
    
    Returns:
    --------
    str: Classification of the person based on age.
    
    >>> age_classify(5)
    'Child'
    >>> age_classify(15)
    'Teenager'
    >>> age_classify(25)
    'Adult'
    >>> age_classify(70)
    'Elderly'
    '''
    if age < 0:
        return "Invalid age"
    elif age <= 12:
        return "Child"
    elif age <= 19:
        return "Teenager"
    elif age <= 64:
        return "Adult"
    else:
        return "Elderly"
```

## Exercise 7

Create a function that receives three parameters: the given name and the two family names of a person, and returns the full name with both family names. It should be callable with two arguments (given name and first family name) or with three (given name and two family names).

```python
def join_names(name: str, surname1: str, surname2: str = "") -> str:
    '''
    Join a name and one or two surnames into a full name.
    
    Parameters:
    -----------
    name (str): First name.
    surname (str): First surname.
    surname2 (str, optional): Second surname. Defaults to "".
    
    Returns:
    --------
    str: Full name.
    
    >>> join_names("John", "Doe")
    'John Doe'
    >>> join_names("Jane", "Smith", "Johnson")
    'Jane Smith Johnson'
    '''
    if surname2:
        return name + " " + surname1 + " " + surname2
    else:
        return name + " " + surname1
```

```python
def join_names(name: str, surname1: str, surname2: str = "") -> str:
    '''
    Join a name and one or two surnames into a full name.
    
    Parameters:
    -----------
    name (str): First name.
    surname (str): First surname.
    surname2 (str, optional): Second surname. Defaults to "".
    
    Returns:
    --------
    str: Full name.
    
    >>> join_names("John", "Doe")
    'John Doe'
    >>> join_names("Jane", "Smith", "Johnson")
    'Jane Smith Johnson'
    '''
    if surname2:
        return f"{name} {surname1} {surname2}"
    else:
        return f"{name} {surname1}"
```
